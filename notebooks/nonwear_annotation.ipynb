# Manual nonwear annotation

Run the cell below to open the local annotation app in your browser.
It uses `configs/nonwear_annotation.local.toml` and resumes saved labels.
Keep this kernel running while reviewing. Re-running the cell reuses the running app.
See `docs/nonwear_annotation.md` for plot controls, saving, and exports.


```python
from pathlib import Path
import subprocess
import sys
import time
import webbrowser
import urllib.request
import urllib.error
from IPython.display import display, HTML

PROJECT_ROOT = next(p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
                    if (p / "scripts" / "annotate_lower_back_nonwear.py").is_file())
ANNOTATION_PORT = 8765
ANNOTATION_URL = f"http://127.0.0.1:{ANNOTATION_PORT}"
if "annotation_process" not in globals() or annotation_process.poll() is not None:
    annotation_process = subprocess.Popen(
        [str(PROJECT_ROOT / ".venv" / "bin" / "python"),
         str(PROJECT_ROOT / "scripts" / "annotate_lower_back_nonwear.py"),
         "--port", str(ANNOTATION_PORT), "--no-browser"], cwd=PROJECT_ROOT)
    for _ in range(100):
        if annotation_process.poll() is not None:
            raise RuntimeError("The annotation app did not start. Check the printed error or run the terminal launcher.")
        try:
            with urllib.request.urlopen(ANNOTATION_URL + "/api/session", timeout=1):
                break
        except (urllib.error.URLError, TimeoutError):
            time.sleep(0.1)
    else:
        raise RuntimeError("The app is still starting. Use the terminal launcher to inspect its progress.")
webbrowser.open(ANNOTATION_URL)
display(HTML(f'<a href="{ANNOTATION_URL}" target="_blank">Open nonwear annotation</a>'))
```
